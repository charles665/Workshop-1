# Initial data profiling
Focused profiling required by Workshop-1; this notebook does not transform or overwrite the source.

```python
from pathlib import Path
import pandas as pd

source_path = Path('../data/raw/candidates.csv')
df = pd.read_csv(source_path, sep=';')
df.shape
```

```python
display(pd.DataFrame({'dtype': df.dtypes.astype(str), 'missing_values': df.isna().sum(), 'unique_values': df.nunique()}))
print('Exact duplicate rows:', df.duplicated().sum())
display(df.head())
```

```python
application_date = pd.to_datetime(df['Application Date'], errors='coerce')
print('Invalid dates:', application_date.isna().sum())
print('Date range:', application_date.min().date(), 'to', application_date.max().date())
display(df[['YOE', 'Code Challenge Score', 'Technical Interview Score']].describe())
```

```python
for column in ['Country', 'Seniority', 'Technology']:
    print(f'\n{column} ({df[column].nunique()} unique)')
    display(df[column].value_counts().head(15).to_frame('count'))

hired = ((df['Code Challenge Score'] >= 7) & (df['Technical Interview Score'] >= 7))
print('Hired:', int(hired.sum()), f'({hired.mean():.2%})')
```
